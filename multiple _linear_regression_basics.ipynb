```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn import linear_model as lm
```

```python
df=pd.read_csv("house price.csv")
df
```

```python
import math as m
mid=m.floor(df.beds.median())
df.beds=df.beds.fillna(mid)
df
```

```python
reg=lm.LinearRegression()
reg.fit(df[["beds","age","area"]],df.price)
reg.predict(pd.DataFrame({"beds":[19],
                          "age":19,
                          "area":10000}))
```
