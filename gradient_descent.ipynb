```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array([1,2,3,4,5])
y=np.array([10,20,30,40,50])
```

```python
def gradient_descent(x,y):
    m_current=b_current=0
    iterations=1000
    n=len(x)
    m_history =[]
    b_history=[]
    learning_rate=0.01
    for i in range(iterations):
        y_pred=(m_current*x)+b_current
        md=-(2/n)*sum(x*(y-y_pred))
        bd=-(2/n)*sum(y-y_pred)
        cost_function=(1/n)*sum(np.pow(y-y_pred,2))
        m_current=m_current-learning_rate*md
        b_current-=learning_rate*bd
        m_history.append(m_current)
        b_history.append(b_current)
        print(f"m:{m_current}")
        print(f"b:{b_current}")
        print(f"iteration:{i}")
        print(f"m_current*x+b_current:{m_current*x+b_current}")
        print(f"cost:{cost_function}")
        #plotting the graph between values  of x and m*x+b
        plt.plot(x,y_pred)
        plt.xlabel("value of x",color='blue')
        plt.ylabel("y=mx+b",color='red')
        plt.title("x vs predicted y")
```

```python
gradient_descent(x,y)
```
