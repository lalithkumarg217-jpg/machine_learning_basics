```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
from sklearn import linear_model as lm
```

```python
df=pd.DataFrame({"price":(40000,839478,992884),
                "area":(400,864,978)})
```

```python
plt.scatter(df.price,df.area)
plt.xlabel("price",color="red")
plt.ylabel("area",color="green")
plt.title("price vs area",color="purple")
plt.show()
```

```python
reg=lm.LinearRegression()
reg.fit(df[["price"]],df["area"])
reg.predict(pd.DataFrame({"price":[1234567]}))
```
